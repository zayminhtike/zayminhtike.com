# Data Visualization with Seaborn

Seaborn is a powerful visualization library built on top of Matplotlib. It's designed to make creating common statistical plots easy and aesthetically pleasing.

In this notebook, we'll explore the famous **Titanic dataset** to learn how to use visualization for EDA




---

## The Titanic Dataset

This dataset describes the survival status of individual passengers on the Titanic. The titanic data frame does not contain information from the crew, but it does contain actual ages of (some of) the passengers.

### Variables
 - survived - Survival (Yes, No)
 - pclass - Passenger Class (1st, 2nd, 3rd)
 - sex - Sex (female, male)
 - age - Age in years
 - sibsp - Number of Siblings/Spouses Aboard
 - parch - Number of Parents/Children Aboard
 - fare - Passenger Fare
 - embark_town - Port of Embarkation (Cherbourg, Queenstown, Southampton)

## Setup and Loading Data

First, we need to import the necessary libraries. We'll need `seaborn` for plotting, `matplotlib.pyplot` for customizing our plots (like adding titles), and `pandas` for data manipulation, though Seaborn can load the dataset for us directly.

Seaborn comes with several built-in datasets, including the Titanic dataset. Let's load it and take a quick look.

```python
# Import the libraries
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Set a nice style for the plots
sns.set_style('whitegrid')
```

Seaborn's Five Preset Styles
Here are the five primary styles available:

**darkgrid** (Default): This style features a dark gray background with prominent white grid lines. It's great for helping users trace data points across the plot.

**whitegrid**: Similar to darkgrid, but with a clean white background and gray grid lines. This is a very popular choice for its readability.

**dark**: A simple dark gray background with no grid lines. This style can make colorful plot elements stand out.

**white**: A classic, minimalist style with a plain white background and no grid lines. It looks very similar to a standard Matplotlib plot.

**ticks**: This style has a white background and adds small tick marks on the axes, which can be useful for seeing the exact location of data points without the clutter of a full grid.

```python
# Load the Titanic dataset
titanic = sns.load_dataset('titanic')

# Display the first 5 rows of the dataset to see what it looks like
print("First 5 rows of the Titanic dataset:")
titanic.head()
```

Let's also check for missing values, as this is important for plotting. We'll focus on the `age` column, which has some missing data. Seaborn is generally smart enough to ignore these missing values automatically.

```python
# Get a summary of the data, including missing values
print("\nDataset Information:")
titanic.info()
```

## Visualizing Distributions of columns/features

### Histogram

A **histogram** is used to see the distribution of a single numerical variable. It groups data into "bins" and shows how many data points fall into each bin.

Let's visualize the distribution of passenger ages on the Titanic. We'll use the `sns.histplot()` function.

```python
# Create a histogram of the 'age' column
plt.figure(figsize=(10, 6)) # Creating a container for the plot
sns.set_style('dark')
sns.histplot(data=titanic,
             x='age',
             bins=30,
             kde= False  # kde=False to not show the density line yet
             )
# Add a title and labels for clarity
plt.title('Distribution of Passenger Ages on the Titanic', fontsize=16)
plt.xlabel('Age', fontsize=12)
plt.ylabel('Number of Passengers', fontsize=12)

# Show the plot
plt.show()
```

### Density Plot (KDE Plot)

A **Kernel Density Estimate (KDE) plot**, or density plot, is like a smoothed version of a histogram. It helps visualize the probability density of a continuous variable.

Let's create a density plot for `age` using `sns.kdeplot()`.

```python
# Create a density plot of the 'age' column
plt.figure(figsize=(10, 6))
sns.kdeplot(data=titanic, x='age', fill=True) # fill=True shades the area under the curve

# Add a title and labels
plt.title('Density of Passenger Ages on the Titanic', fontsize=16)
plt.xlabel('Age', fontsize=12)
plt.ylabel('Density', fontsize=12)

plt.show()
```

### Bar Chart

A **bar chart** is perfect for comparing quantities across different categories. A common use case is to show the counts of items in each category.

Let's see how many passengers were in each class (First, Second, Third). We'll use `sns.countplot()` which is a special type of bar chart that automatically counts the occurrences of each category.

```python
# Create a bar chart showing the count of passengers in each class
plt.figure(figsize=(10, 6))
sns.countplot(data=titanic,
              x='class',
              hue = 'class',
              palette='viridis',
              legend= False,
              )

# Add a title and labels
plt.title('Number of Passengers by Class', fontsize=16)
plt.xlabel('Passenger Class', fontsize=12)
plt.ylabel('Number of Passengers', fontsize=12)

plt.show()
```

### Box Plot

A **box plot** (or box-and-whisker plot) is excellent for visualizing the distribution of a numerical variable across different categories. It displays the five-number summary: minimum, first quartile (Q1), median, third quartile (Q3), and maximum.

Let's compare the age distribution across the different passenger classes using `sns.boxplot()`.

```python
# Create a box plot to show age distribution by class
plt.figure(figsize=(10, 6))
sns.boxplot(data=titanic,
            x='class',
            hue = 'class',
            y='age',
            palette='pastel',
            legend = False
            )

# Add a title and labels
plt.title('Age Distribution by Passenger Class', fontsize=16)
plt.xlabel('Passenger Class', fontsize=12)
plt.ylabel('Age', fontsize=12)

plt.show()
```

**How to read a box plot:**
* The **box** shows the interquartile range (IQR), from Q1 to Q3.
* The **line inside the box** is the median (the 50th percentile).
* The **whiskers** extend to show the range of the data, typically 1.5 times the IQR.
* The **dots** outside the whiskers are potential outliers.

### Violin Plot

A **violin plot** is a powerful combination of a box plot and a density plot. It shows the same summary statistics as a box plot, but also includes the probability density of the data at different values.

Let's use a violin plot to compare the age distribution for male and female passengers. We can use `sns.violinplot()`.

```python
# Create a violin plot to show age distribution by sex
plt.figure(figsize=(10, 6))
sns.violinplot(data=titanic, x='sex', y='age', palette='muted')

# Add a title and labels
plt.title('Age Distribution by Sex', fontsize=16)
plt.xlabel('Sex', fontsize=12)
plt.ylabel('Age', fontsize=12)

plt.show()
```

**Observation**: The shape of the violin (the density plot) confirms that the most common age for both males and females was in the 20-30 range. The white dot inside is the median, and the thick black bar is the interquartile range, just like in a box plot. This plot gives us a much richer understanding of the distribution than a simple box plot would.

## Visualizing Multiple Columns/Features to explore correlations

### Scatterplot

A **scatterplot** is perfect for investigating the relationship between two continuous (numerical) variables. Let's see if there's a connection between a passenger's **age** and the **fare** they paid.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=titanic,
                x='age',
                y='fare',
                hue='survived',
                legend = True
                )
plt.title('Passenger Age vs. Fare Paid')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

### Heatmap

A **heatmap** is a great way to visualize the correlation between multiple numerical variables at once. A correlation matrix shows how strongly variables are related to each other, with values ranging from -1 (perfect negative correlation) to +1 (perfect positive correlation).

```python
# Select only numerical columns for the correlation matrix
numerical_cols = titanic.select_dtypes(include=['float64', 'int64'])

plt.figure(figsize=(10, 8))
sns.heatmap(numerical_cols.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

### Stacked vs. Grouped Bar Charts

Bar charts are ideal for comparing categorical data. Let's compare survival rates based on passenger class (`pclass`) using both stacked and grouped bar charts.

### Stacked Bar Chart

A **stacked bar chart** shows the composition of each category. It's good for understanding the proportions within each bar. We first need to reshape our data using Pandas' `crosstab` function.

```python
import pandas as pd

# Create a cross-tabulation of Pclass and Survived. Pclass will be rows and Survived will be columns
class_survival = pd.crosstab(titanic['pclass'], titanic['survived'])

# Create the stacked bar chart
class_survival.plot(kind='bar',
                    stacked=True,
                    figsize=(10, 6)
                    )
plt.title('Survival Count by Passenger Class (Stacked)')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

### Grouped Bar Chart

A **grouped bar chart** places bars for different categories side-by-side, making direct comparisons easier. Seaborn's `countplot` with the `hue` parameter is perfect for this.

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=titanic,
              x='pclass',
              hue='survived'
              )
plt.title('Survival Count by Passenger Class (Grouped)')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

### Visualizing a Categorical and a Numerical Feature

There are several excellent ways to visualize the distribution of a numerical variable (like `age`) across different categories (like `survived`).

### Strip Plot and Swarm Plot

These plots show every single data point, which is useful for seeing the underlying distribution.

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
```

The selected line `fig, axes = plt.subplots(1, 2, figsize=(16, 6))` uses the `subplots()` function from the matplotlib.pyplot library (which we imported as plt) to create a figure and a set of subplots.

`plt.subplots()` is a convenient function for creating figures and axes together.
1, 2 specifies that we want a grid of 1 row and 2 columns of subplots.
`figsize=(16, 6)` sets the size of the entire figure in inches. This makes the overall plot wider (16 inches) than it is tall (6 inches), providing space for two plots side-by-side.

The function returns two things:

`fig`: This is the Figure object, which is the overall container for the plot. You can use this to set things like the main title for the entire figure.

`axes`: This is an array containing the Axes objects for each subplot. Since we asked for 1 row and 2 columns, axes will be a 1D array with two elements, axes[0] for the first subplot and axes[1] for the second subplot. We can then pass these axes objects to plotting functions (like sns.stripplot and sns.swarmplot in the next lines) to tell them which subplot to draw on.

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
# Strip Plot
sns.stripplot(ax=axes[0],
              data=titanic,
              x='survived',
              y='age')

axes[0].set_title('Strip Plot of Age by Survival Status')

# Swarm Plot
sns.swarmplot(ax=axes[1],
              data=titanic,
              x='survived',
              y='age')

axes[1].set_title('Swarm Plot of Age by Survival Status')

plt.show()
```

**Interpretation:** The **strip plot** shows all points but suffers from overplotting where points overlap. The **swarm plot** is an improvement, adjusting points to avoid overlap. Both show that the age distribution for survivors and non-survivors is quite similar, though there are slightly more young children among the survivors.

### Density Plot (KDE)

A **density plot** (or Kernel Density Estimate) visualizes the probability density of a continuous variable. It's like a smoothed histogram.

```python
plt.figure(figsize=(10, 6))
sns.kdeplot(data=titanic, x='age', hue='survived', fill=True, common_norm=False)
plt.title('Density Plot of Age by Survival Status')
plt.show()
```

**Interpretation:** The peak for survivors is noticeably higher for very young children (age 0-5), confirming they had a higher survival rate.

### Violin Plot and Box Plot

These plots are excellent for summarizing distributions.

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Violin Plot
sns.violinplot(ax=axes[0], data=titanic, x='sex', y='age', hue='survived', split=True)
axes[0].set_title('Violin Plot of Age by Sex and Survival')

# Box Plot
sns.boxplot(ax=axes[1], data=titanic, x='sex', y='age', hue='survived')
axes[1].set_title('Box Plot of Age by Sex and Survival')

plt.show()
```

**Interpretation:**
* The **violin plot** combines a box plot with a density plot. The width of the violin shows the frequency of data points at that age. We can clearly see the "bump" for surviving children, especially males.
* The **box plot** summarizes the distribution with quartiles. The box represents the interquartile range (IQR), the line inside is the median, and the whiskers show the range of the data. It clearly shows the median age for surviving females was slightly lower than for those who didn't. Both plots confirm that survival was not strongly dependent on age for adults.

### Faceting: Visualizing More Than Two Features

**Faceting** is the powerful technique of creating multiple subplots (a "grid") based on the levels of one or more categorical variables. This allows you to visualize complex relationships involving 3, 4, or even more variables. Seaborn's `catplot` is a great tool for this.

Let's visualize the relationship between **survival**, **age**, and **sex**, faceted by **passenger class**.

```python
sns.catplot(data=titanic,
            x='sex',
            y='age',
            hue='survived',
            col='pclass',
            kind='violin',
            split=True)

plt.suptitle('Age Distribution by Sex, Survival, and Passenger Class', y=1.02) # y adjusts title position
plt.show()
```

#### Interpretation

This single command generates a rich visualization showing four different variables:
* `x='sex'`: Male vs. Female
* `y='age'`: Distribution of Age
* `hue='survived'`: Survived vs. Not Survived
* `col='pclass'`: A separate column for each passenger class

From this faceted plot, we can draw more nuanced conclusions:
* In **1st Class**, a large proportion of women of all ages survived. Most male children survived, but a significant number of adult men did not.
* In **2nd Class**, almost all women and children survived. The vast majority of men did not.
* In **3rd Class**, survival was poor for both sexes, though women still fared better than men. We can also see that many children in 3rd class, both male and female, did not survive.
