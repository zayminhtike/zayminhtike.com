### Data Preparation

```python
# Import packages
import pandas as pd        # Import pandas for data manipulation and analysis
import numpy as np         # Import numpy for numerical operations
import matplotlib.pyplot as plt  # Import matplotlib for plotting graphs

from sklearn.linear_model import LogisticRegression  # Import LogisticRegression for building a classification model
```

```python
# Load the Wisconsin Breast Cancer dataset
WBCD = pd.read_csv("WisconsinBreastCancerDatabase.csv")
# Read the CSV file containing breast cancer data into a DataFrame
```

```python
# Logistic Regression with Scikit-learn needs binary input in the form of 0s and 1s
# A common data preparation step is 'hot encoding', or converting a categorical column into 0s and 1s
# If there are multiple categories, then multiple 0/1 columns are created.

# Convert Diagnosis to 0 and 1
WBCD['Diagnosis_encoded'] = WBCD['Diagnosis'].map({'B': 0, 'M': 1})

# Display the DataFrame to show the loaded and modified data
WBCD
```

### Setting up the Scikit Learn Workflow: Define -> Initialize -> Fit -> Predict -> Evaluate

#### Define

```python
# Store relevant columns as variables
X = WBCD[['Radius mean']].values.reshape(-1, 1)
# This line selects the 'Radius mean' column from your DataFrame WBCD, converts
# it into a NumPy array using .values, and then reshapes it into a 2D array with
# one column using .reshape(-1, 1). This reshaping is necessary because
# scikit-learn models expect the input features (X) to be in this format.

y = WBCD['Diagnosis_encoded'].values.astype(int)
# Similarly, this line selects the 'Diagnosis_encoded' column, converts it to a
# 1D NumPy array, and then converts the data type of the array to
# integers using .astype(int). This is done because the target variable (y) for
# classification models in scikit-learn should be integers.
```

```python
y
```

#### Initialize

```python
logisticModel = LogisticRegression()
# This line creates an instance of the LogisticRegression model from the
# scikit-learn library. This is your classification model object.

```

#### Fit

```python
logisticModel.fit(X, y)
# This line trains the logistic regression model. It takes your feature data (X)
# and your target variable (y) as input.
# The model learns the relationship
# between the 'Radius mean' and the 'Diagnosis' based on this data.
```

#### Predict

```python
yPredicted = logisticModel.predict(X)
```

```python
yPredicted
```

#### Evaluate

```python
logisticModel.score(X,y)
```

Plotting Model along with raw data

```python

# Graph logistic regression probabilities
plt.scatter(X, y)

xDelta = np.linspace(X.min(), X.max(), 1000)
# This generates a sequence of 10,000 evenly spaced values between the minimum
# and maximum of your 'Radius mean' data. This is used to create a smooth curve
# for the logistic regression line.

yPredicted = logisticModel.predict(X).reshape(-1, 1).astype(int)
# This line uses the trained logistic regression model to predict the diagnosis
# (0 or 1) for each data point in your original dataset (X).

yDeltaProb = logisticModel.predict_proba(xDelta.reshape(-1, 1))[:, 1]
# This uses the model to predict the probability of a tumor being
# malignant (the second column, indexed by [:, 1]) for each of
# the xDelta values.

plt.plot(xDelta, yDeltaProb, color='red')

plt.xlabel('Radius', fontsize=14);

plt.ylabel('Probability of malignant tumor', fontsize=14);
```

### Creating Regression Table (Optional)
The `statsmodels` library can be used to display the results of model fit in the form of a table

```python
import statsmodels.api as sm
# This line imports the statsmodels.api module, which is commonly used for
# statistical modeling in Python. We give it the alias sm for convenience.

# Add a constant (intercept) to the feature data
X_sm = sm.add_constant(X)
# This is the key line you selected. In linear and logistic regression, an
# intercept term (also called a constant) is usually included in the model. This
# constant represents the expected value of the target variable when all the
# predictor variables are zero. The sm.add_constant() function takes your
# feature data X (which is currently just the 'Radius mean' column) and adds a
# new column of ones to it. This new column represents the constant term in the
# regression equation. The resulting array, X_sm, now includes both your 'Radius
# mean' data and a column of ones for the intercept.

```

```python
X_sm
```

```python
# Create and fit the logistic regression model using statsmodels
logisticModel_sm = sm.Logit(y, X_sm)
# This line creates a Logit model object from the statsmodels library. You
# provide the target variable (y) and the feature data with the added constant
# (X_sm) to this function.

result = logisticModel_sm.fit()
# This line fits the logistic regression model to your data. The fit() method
# estimates the model parameters (coefficients and intercept) based on the
# provided data. The results of the fitting process are stored in the result
# object, which contains various statistical summaries.

```

In the output of the above cell, the "Current function value" represents the value of the "Loss Function" that the method arrived at after optimization.
In the case of logistic regression, the loss function is defined as the negative log-likelihood function.

During the fitting process, statsmodels uses an iterative optimization algorithm (like gradient descent) to find the model coefficients that minimize the negative log-likelihood. The negative log-likelihood is a measure of how well the model fits the data; a lower value indicates a better fit.

The "Current function value" you see is the value of this function at the point where the optimization terminated successfully. The goal of the optimization is to find the minimum possible value for this function.

```python
# Display the regression table summary
print(result.summary())
```
