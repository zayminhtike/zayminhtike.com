# Building a single Perceptron model

### Import Libraries

```python
# This section loads all the Python libraries we need for our perceptron model

# pandas (pd): Used for working with data in table format (DataFrames)
import pandas as pd

# numpy (np): Used for mathematical operations and working with arrays
import numpy as np

# seaborn (sns): Used for creating beautiful statistical visualizations
import seaborn as sns

# Perceptron: The main machine learning algorithm we'll use (a simple neural network)
from sklearn.linear_model import Perceptron

# train_test_split: Helps us split our data into training and testing sets
from sklearn.model_selection import train_test_split

# StandardScaler: Normalizes our data so all features are on the same scale
from sklearn.preprocessing import StandardScaler

# accuracy_score: Calculates how many predictions were correct
# classification_report: Shows detailed performance metrics (precision, recall, f1-score)
# confusion_matrix: Shows where the model made correct/incorrect predictions
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

### Data Import and Pre-Processing

```python

# The Haberman dataset contains survival data for patients who had breast cancer surgery
# Features: Age of patient, Year of operation, Number of positive axillary nodes detected
# Target: Survival status (1 = survived 5+ years, 0 = died within 5 years)

# Load the CSV file into a pandas DataFrame called 'haberman'
haberman = pd.read_csv('haberman.csv')
```

```python
haberman
```

### Scikit-Learn Workflow

#### DEFINE

```python

# In machine learning, we separate our data into:
#   X = input features (the data the model uses to make predictions)
#   y = target/label (what we're trying to predict)

# X contains the three input features: Age, Year, and Nodes
X = haberman[['Age', 'Year', 'Nodes']]

# y contains the target variable: Survived (1 or 0)
y = haberman[["Survived"]]
```

```python

# Feature scaling is important for the Perceptron algorithm!
# It transforms all features to have the same scale (mean=0, standard deviation=1)
# This prevents features with large values from dominating the learning process

# Create a StandardScaler object (this will learn the mean and std of our data)
scaler = StandardScaler()

# Fit the scaler to X and transform it (normalize the features)
# The result is converted back to a DataFrame for easier viewing
X_scaled = pd.DataFrame(scaler.fit_transform(X), columns=['Age', 'Year', 'Nodes'])
```

```python
X_scaled
```

```python
# ========================================
# Split Data into Training and Testing Sets
# ========================================
# We split our data into two parts:
#   - Training set (80%): Used to train/teach the model
#   - Testing set (20%): Used to evaluate how well the model performs on unseen data
# This helps us know if the model can generalize to new data

# train_test_split splits X_scaled and y into 4 parts:
# X_train, X_test: The scaled features for training and testing
# y_train, y_test: The target values for training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=123
)
```

Convert *y_train* and *y_test* into 1D NumPy array

```python
y_train = np.ravel(y_train)
y_test = np.ravel(y_test)
```

#### INITIALIZE

```python
# A Perceptron is a simple type of artificial neural network
# It learns by adjusting weights based on errors in its predictions

# Create a Perceptron classifier with specific parameters:
#   tol=0.00001: Stop training when improvement is smaller than this value
#   eta0=0.1: Learning rate (how big each weight adjustment is)
#   max_iter=20000: Maximum number of training iterations
clf = Perceptron(tol=0.00001, eta0=0.1, max_iter=20000)
```

#### FIT/TRAIN MODEL

```python
# Train (fit) the model using the training data
# np.ravel(y_train) converts y_train from 2D to 1D array format
clf.fit(X_train, y_train)
```

#### PREDICT

```python
# ========================================
# Make Predictions on Test Data
# ========================================
# Now that the model is trained, we use it to make predictions on the test set
# The model has never seen this test data before, so this shows how well it generalizes

# Use the trained model to predict survival for the test data
# clf.predict() takes the test features and returns predictions (0 or 1)
y_pred = clf.predict(X_test)
```

#### EVALUATE

```python
# ========================================
# Display Model Predictions
# ========================================
# Let's look at what the model predicted for each test sample
# This shows an array of 0s and 1s (the model's predictions)

y_pred
```

```python
# ========================================
# Display Actual Values
# ========================================
# Let's compare the predictions with the actual survival values from the test set
# This shows the true labels (what actually happened to each patient)

y_test
```

```python
# ========================================
# Display Confusion Matrix as a Table
# ========================================
# This shows the confusion matrix in a clean table format
# Rows = Predicted values, Columns = Actual values

import pandas as pd

# Create the confusion matrix
cm = confusion_matrix(y_pred, y_test)

# Convert to DataFrame for better visualization
cm_df = pd.DataFrame(
    cm,
    index=['Predicted 0', 'Predicted 1'],
    columns=['Actual 0', 'Actual 1']
)

print("Confusion Matrix:")
print(cm_df)
print("\nInterpretation:")
print(f"True Negatives (TN): {cm[0,0]} - Correctly predicted 0")
print(f"False Positives (FP): {cm[0,1]} - Incorrectly predicted 1 (actually 0)")
print(f"False Negatives (FN): {cm[1,0]} - Incorrectly predicted 0 (actually 1)")
print(f"True Positives (TP): {cm[1,1]} - Correctly predicted 1")
```

```python
# ========================================
# Calculate Accuracy Score
# ========================================
# Accuracy tells us what percentage of predictions were correct
# It's calculated as: (correct predictions) / (total predictions)
# A score of 1.0 = 100% accurate, 0.5 = 50% accurate, etc.

accuracy_score(y_pred, y_test)
```

```python
# ========================================
# Display Model Weights and Intercept
# ========================================
# The perceptron learns weights for each feature and an intercept (bias)
# These values determine how the model makes its decisions

# Display the weights (coefficients) for each feature
# Each weight shows how important that feature is for the prediction
print(clf.coef_)

# Display the intercept (bias term)
# This is the starting point before considering the features
print(clf.intercept_)
```

```python
# compute accuracy on training data
y_train_pred = clf.predict(X_train)
training_data_accuracy = accuracy_score(y_train_pred, y_train)
training_data_accuracy
```

```python
# ========================================
# Display Classification Report
# ========================================
# The classification report provides detailed performance metrics for each class:
#   - Precision: Of all positive predictions, how many were actually correct?
#   - Recall: Of all actual positives, how many did we correctly identify?
#   - F1-score: Harmonic mean of precision and recall (balanced measure)
#   - Support: Number of actual occurrences of each class in the test set

# Print a detailed report comparing predictions to actual values
print(classification_report(y_pred, y_test))
```

# Multilayer Perceptron Model to predict home prices

This section demonstrates how to build and train a Multilayer Perceptron (MLP) neural network to predict house prices using the homes dataset.

### Import Libraries


```python
# Import numpy for numerical operations and array manipulation
import numpy as np

# Import pandas for data loading and manipulation
import pandas as pd

# Import matplotlib.pyplot for creating visualizations
import matplotlib.pyplot as plt

# Import seaborn for enhanced statistical visualizations
import seaborn as sns

# Import train_test_split to divide data into training and testing sets
from sklearn.model_selection import train_test_split

# Import MLPRegressor, the multilayer perceptron neural network for regression
from sklearn.neural_network import MLPRegressor

# Import mean_squared_error to evaluate model performance
from sklearn.metrics import mean_squared_error

# Import matplotlib inline backend for displaying plots in notebook
import matplotlib_inline.backend_inline

# Set matplotlib to use SVG format for higher quality plots
matplotlib_inline.backend_inline.set_matplotlib_formats('svg')
```

### Data Loading and Pre-Processing



```python
# Read the CSV file 'homes.csv' using pandas and store it in a DataFrame
homes = pd.read_csv('homes.csv')

# Display the first few rows of the dataset to verify successful loading
homes
```

### Scikit-Learn Workflow

#### DEFINE

Prepare the data by selecting features and splitting into training and test sets.

```python
# Select the input features: 'Bed' (number of bedrooms) and 'Floor' (floor area in thousands of sq ft)
X = homes[['Bed', 'Floor']]

# Select the output target variable: 'Price' (house price in thousands of dollars)
y = homes[['Price']]
```

```python
# Split the data into training and testing sets
# X_train, X_test: training and testing input features
# y_train, y_test: training and testing target values
# np.ravel(y): flatten the y array to 1D
# random_state=123: set seed for reproducibility
# Default is 75% Training and 25% Testing
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=123)
y_train = np.ravel(y_train)
y_test = np.ravel(y_test)
```

#### INITIALIZE

```python
# Create an MLPRegressor (Multilayer Perceptron for regression)
mlpReg_train = MLPRegressor(
    random_state=1,           # Set random seed for reproducibility
    max_iter=500000,   # Maximum number of training iterations
    tol = 1e-4     ,  # Controls the minimum change in the loss between iterations needed to continue training.
    hidden_layer_sizes=[1],# Neural network architecture: 1 hidden layer with 1 neuron
    activation='relu'  # Activation function: Rectified Linear Unit (ReLU). This is the default for MLPRegressor
    )
```

#### FIT/TRAIN MODEL


```python
# Train the model using the training data
# X_train: input features, np.ravel(y_train): flattened target values
mlpReg_train.fit(X_train, y_train)
```

#### PREDICT

Use the trained model to make predictions on new data.

```python
# Use the trained model to predict the price of a house
# Example: Predict price for a house with 5 bedrooms and 2.896 thousand sq ft floor area
# The prediction returns the estimated price in thousands of dollars
mlpReg_train.predict([[5, 2.896]])
```

#### EVALUATE

Visualize the training loss curve to evaluate model performance during training.

```python
# Create a figure with 1 subplot (1 row, 1 column)
f, ax = plt.subplots(1, 1)

# Create a line plot using seaborn
sns.lineplot(
    x=range(len(mlpReg_train.loss_curve_)),  # X-axis: training iteration numbers (epochs)
    y=mlpReg_train.loss_curve_,               # Y-axis: loss values at each iteration
    label='Training'                          # Label for the legend
)

# Set the x-axis label with font size 14
ax.set_xlabel('Epochs', fontsize=14)

# Set the y-axis label with font size 14
ax.set_ylabel('Loss', fontsize=14)
```

```python
# Compare the final loss between train and test sets
print(mlpReg_train.loss_)
print(
    mean_squared_error(y_test, mlpReg_train.predict(X_test)) / 2
)  # division by 2 to get squared error to match squared error.
```

```python
# Obtains the final weights and biases
print(mlpReg_train.coefs_)
print(mlpReg_train.intercepts_)
```
