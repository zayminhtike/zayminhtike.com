# Unsupervised Learning

## Kmeans Clustering

#### Import packages and functions

```python
# Import packages and functions
import matplotlib.pyplot as plt  # Import matplotlib for creating visualizations
import seaborn as sns  # Import seaborn for enhanced statistical visualizations
import pandas as pd  # Import pandas for data manipulation and analysis

from sklearn.cluster import KMeans  # Import KMeans clustering algorithm from scikit-learn
```

#### Load dataset

```python

geyser = pd.read_csv('oldfaithful.csv')  # Read the CSV file containing Old Faithful geyser data into a pandas DataFrame
geyser  # Display the DataFrame to view its contents
```

#### Visual exploration

```python

p = sns.scatterplot(data=geyser, x='Eruption', y='Waiting')  # Create a scatter plot with Eruption time on x-axis and Waiting time on y-axis
p.set_xlabel('Eruption time (min)', fontsize=14);  # Set the x-axis label with font size 14
p.set_ylabel('Waiting time (min)', fontsize=14);  # Set the y-axis label with font size 14
```

#### Initialize K-Means model

```python
# Initialize a k-means model with k=2
kmModel = KMeans(n_clusters=2)  # Create a KMeans object with 2 clusters
```

#### Fit K-Means Model

```python
# Fit the model
kmModel = kmModel.fit(geyser)  # Train the KMeans model on the geyser dataset
```

#### Create the cluster assignments

```python

clusters = kmModel.fit_predict(geyser[['Eruption', 'Waiting']])  # Predict cluster labels for each data point using the 'Eruption' and 'Waiting' columns

# View the clusters for the first five instances
clusters[0:5]  # Display the cluster assignments for the first 5 data points
```

#### Plot clusters

```python
# Save the cluster centroids
centroids = kmModel.cluster_centers_  # Extract the coordinates of the cluster centers
centroids[1]  # Display the centroid coordinates of cluster 1
```

```python
p = sns.scatterplot(  # Create a scatter plot using seaborn
    data=geyser, x='Eruption', y='Waiting', hue=clusters, style=clusters  # Plot with Eruption on x-axis, Waiting on y-axis, colored and styled by cluster assignment
)
p.set_xlabel('Eruption time (min)', fontsize=14);  # Set the x-axis label with font size 14
p.set_ylabel('Waiting time (min)', fontsize=14);  # Set the y-axis label with font size 14

# Add centroid for cluster 0
plt.scatter(x=centroids[0, 0], y=centroids[0, 1], c='black')  # Plot the centroid of cluster 0 as a black point

# Add centroid for cluster 1
plt.scatter(x=centroids[1, 0], y=centroids[1, 1], c='black', marker='X')  # Plot the centroid of cluster 1 as a black X marker
```

#### Identifying optimal number of clusters

```python
# Fit k-means clustering with k=1,...,5 and save WCSS for each
WCSS = []  # Initialize an empty list to store within-cluster sum of squares values
k = [1, 2, 3, 4, 5]  # Create a list of k values (number of clusters) to test
for j in k:  # Loop through each k value
    kmModel = KMeans(n_clusters=j)  # Create a KMeans model with j clusters
    kmModel = kmModel.fit(geyser)  # Fit the model to the geyser dataset
    WCSS.append(kmModel.inertia_)  # Append the inertia (WCSS) value to the list
```

```python
# Plot the WCSS for each cluster
ax = plt.figure().gca()  # Create a new figure and get the current axes
plt.plot(k, WCSS, '*-')  # Plot k values on x-axis and WCSS values on y-axis with markers and lines
plt.xlabel('Number of clusters (k)', fontsize=14);  # Set the x-axis label with font size 14
plt.ylabel('Within-cluster sum of squares (WCSS)', fontsize=14);  # Set the y-axis label with font size 14
```

## Hierarchical Clustering

#### Load Libraries

```python
# Import pandas library for data manipulation and creating DataFrames
import pandas as pd
# Import numpy for numerical operations and array handling
import numpy as np
# Import matplotlib for creating visualizations and plots
import matplotlib.pyplot as plt
# Import dendrogram and linkage functions from scipy for hierarchical clustering analysis
from scipy.cluster.hierarchy import dendrogram, linkage
# Import squareform function to convert distance matrices between condensed and square format
from scipy.spatial.distance import squareform
```

#### Load Data

```python
# Load the dataset containing cytochrome protein differences between species
# Read the cytochrome CSV file
cytochrome_full = pd.read_csv('cytochrome.csv', header=None)
cytochrome_full.head()
```

Euclidean distance is a common similarity measure in clustering, but not the only option. Researchers in the fields of biology and medicine often study the differences between amino acid sequences in a protein. Cytochrome c, for instance, is a chain of 104 amino acids that plays a major role in programmed cell death and varies between species. The number of differences between sequences of amino acid chains can be used to determine whether two species are closely related.

```python
cytochrome.info()
```

```python
# Read the cytochrome CSV file, skipping first column (index 0) and reading columns 1-13
cytochrome = pd.read_csv('cytochrome.csv', header=None, usecols=range(1, 14))
```

```python
cytochrome.head()
```

```python
# Add labels for each species and save as a data frame
# Create a list containing the names of all species in the dataset
# Each name corresponds to a row/column in the cytochrome distance matrix
species = [
    "Human",      # First species - humans
    "Monkey",     # Second species - monkeys
    "Horse",      # Third species - horses
    "Cow",        # Fourth species - cows
    "Dog",        # Fifth species - dogs
    "Whale",      # Sixth species - whales
    "Rabbit",     # Seventh species - rabbits
    "Kangaroo",   # Eighth species - kangaroos
    "Chicken",    # Ninth species - chickens
    "Penguin",    # Tenth species - penguins
    "Duck",       # Eleventh species - ducks
    "Turtle",     # Twelfth species - turtles
    "Frog"        # Thirteenth species - frogs
]

# Create a DataFrame from the cytochrome data with species names as both row and column labels
# This makes the distance matrix easier to read and interpret
cytochrome_df = pd.DataFrame(cytochrome, index=species, columns=species)
# Display the formatted DataFrame showing distances between all pairs of species
cytochrome_df
```

```python
# Format the data as a distance matrix
# In this case, the data already represents distance between points (species)
# Convert the square cytochrome distance matrix to condensed format using squareform function
# This converts the 13x13 symmetric matrix into a 1D array of unique pairwise distances
differences = squareform(cytochrome)
```

```python
differences
```

```python
# Define a clustering model with single linkage
# Create a hierarchical clustering model using single linkage method
# Single linkage measures distance between clusters using the closest pair of points
clusterModel = linkage(differences, method='single')

# Create the dendrogram
# Generate a dendrogram (tree diagram) to visualize the hierarchical clustering
# Labels parameter adds species names to the leaves, leaf_rotation rotates labels 90 degrees for readability
dendrogram(clusterModel, labels=species, leaf_rotation=90)

# Plot the dendrogram with appropriate labels and formatting
# Set the y-axis label to explain what the height represents (amino acid differences)
plt.ylabel('Amino acid differences', fontsize=14)
# Set y-axis ticks from 0 to 10 with step of 1 for clear scale reading
plt.yticks(np.arange(0, 11, step=1))
# Set the x-axis label to indicate that species are shown on the horizontal axis
plt.xlabel('Species', fontsize=14)
# Set the title of the plot to describe the clustering method used
plt.title('Single linkage clustering', fontsize=16)
# Display the completed dendrogram plot
plt.show()
```

## DBSCAN

#### Import packages and functions

```python

import pandas as pd  # Import pandas for data manipulation and analysis
import matplotlib.pyplot as plt  # Import matplotlib for creating visualizations
import seaborn as sns  # Import seaborn for enhanced statistical visualizations

from sklearn.cluster import DBSCAN  # Import DBSCAN clustering algorithm from scikit-learn
from numpy import where  # Import where function from numpy for conditional array operations
from sklearn.preprocessing import StandardScaler  # Import StandardScaler for feature scaling

```

#### Load Data

```python
# Load the dataset
homes = pd.read_csv('homes.csv')  # Read the CSV file containing homes data into a pandas DataFrame

homes  # Display the DataFrame to view its contents
```

```python
# Create a smaller data frame with two variables: Price and Floor
homes_pf = homes[['Price', 'Floor']]  # Select only the 'Price' and 'Floor' columns from the homes DataFrame
homes_pf.describe()  # Display descriptive statistics (count, mean, std, min, max, etc.) for the selected columns
```

#### Standardize data

```python
# Define a scaler to transform values
scaler = StandardScaler()  # Create a StandardScaler object to standardize features by removing the mean and scaling to unit variance

# Apply scaler and view result
homes_scaled = pd.DataFrame(scaler.fit_transform(homes_pf), columns=['Price', 'Floor'])  # Fit the scaler to the data, transform it, and create a new DataFrame with scaled values
homes_scaled.describe()  # Display descriptive statistics for the scaled data
```

#### Initialize DBSCAN model

```python
# Setting a large epsilon will cluster all "middle" values and detect outliers
dbscanModel = DBSCAN(eps=1, min_samples=12)  # Create a DBSCAN object with epsilon=1 (max distance between neighbors) and minimum 12 samples per cluster

# Fit the model
dbscanModel = dbscanModel.fit(homes_scaled)  # Train the DBSCAN model on the scaled homes data
```

#### Predict clusters

```python
clusters = dbscanModel.fit_predict(homes_scaled)  # Fit the DBSCAN model and predict cluster labels for each data point in the scaled homes data
clusters = pd.Categorical(clusters)  # Convert the cluster labels to a categorical data type for better handling
clusters  # Display the cluster assignments
```

#### Visualize scaled outliers

```python

p = sns.scatterplot(data=homes_scaled, x='Floor', y='Price', hue=clusters)  # Create a scatter plot with Floor on x-axis, Price on y-axis, colored by cluster assignments
p.set_xlabel('Scaled floor', fontsize=14);  # Set the x-axis label with font size 14
p.set_ylabel('Scaled price', fontsize=14);  # Set the y-axis label with font size 14
```

```python
# Points where the prediction is -1 are considered outliers
outliers_scaled = homes_scaled[clusters == -1]  # Filter the scaled homes data to include only rows where the cluster label is -1 (outliers)
outliers_scaled  # Display the scaled outlier data
```

```python
# Outliers on original scale (price and square footage in thousands)
outliers_unscaled = homes[clusters == -1]  # Filter the original homes data to include only rows where the cluster label is -1 (outliers)
outliers_unscaled  # Display the unscaled outlier data with original price and square footage values
```

#### Visualizing outliers on original scale

```python

p = sns.scatterplot(data=homes, x='Floor', y='Price', hue=clusters)
p.set_xlabel('Floors', fontsize=14);
p.set_ylabel('Price', fontsize=14);
```

## Principal Component Analysis

```python
# Load the pandas package
# Import pandas library for data manipulation and analysis
import pandas as pd
# Import seaborn for statistical data visualization
import seaborn as sns
# Import numpy for numerical operations and array handling
import numpy as np
# Import scipy.stats for statistical functions
import scipy.stats as st
# Import matplotlib.pyplot for creating plots and visualizations
import matplotlib.pyplot as plt
# Import PCA (Principal Component Analysis) class from sklearn for dimensionality reduction
from sklearn.decomposition import PCA
# Import StandardScaler for feature scaling and normalization
from sklearn.preprocessing import StandardScaler
# Import backend_inline to configure matplotlib display in Jupyter notebooks
import matplotlib_inline.backend_inline
```

```python
# Load the rock.csv dataset
# Read the rock dataset CSV file into a pandas DataFrame
# This dataset contains geological measurements of rock samples including area, perimeter, shape, and permeability
rock = pd.read_csv('rock.csv')
```

```python
# Display the rock dataframe
# Show the entire rock dataset to examine its structure and values
# This will display all rows and columns of geological measurements
rock
```

```python
# Display the correlation matrix using a heatmap
# Create a new figure with a 4x4 inch size for the heatmap visualization
plt.figure(figsize=(4, 4))
# Generate a heatmap showing correlations between all rock variables
# rock.corr() calculates correlation coefficients between all numeric columns
# cmap="YlGnBu" sets the color scheme (Yellow-Green-Blue gradient)
# annot=True displays the correlation values on each cell of the heatmap
sns.heatmap(rock.corr(), cmap="YlGnBu", annot=True)
```

```python
# Create a scatter plot using perimeter and area
# Create a new figure with 4x4 inch dimensions for the plot
plt.figure(figsize=(4, 4))
# Plot perimeter (x-axis) against area (y-axis) as scatter points
# This shows the relationship between rock perimeter measurements and area
plt.scatter(rock['Perimeter'], rock['Area'])
# Set the x-axis label to 'Perimeter' with font size 14
plt.xlabel('Perimeter', fontsize=14);
# Set the y-axis label to 'Area' with font size 14
plt.ylabel('Area', fontsize=14);
```

```python
# Create a scatter plot with a linear regression line
# Perform linear regression analysis on the relationship between rock perimeter and area
# st.linregress returns slope, intercept, and other regression statistics
model = st.linregress(rock['Perimeter'], rock['Area'])
# Create a new figure with 4x4 inch dimensions
plt.figure(figsize=(4, 4))
# Plot the actual data points as a scatter plot
plt.scatter(rock['Perimeter'], rock['Area'])
# Create an array of x values from 0 to 5000 with 10000 evenly spaced points for the regression line
x = np.linspace(0, 5000, 10000)
# Calculate corresponding y values using the linear regression equation: y = slope*x + intercept
# model[0] is the slope, model[1] is the intercept
y = model[0] * x + model[1]
# Plot the regression line in red with 2.5 width
plt.plot(x, y, '-r', linewidth=2.5)
# Set the x-axis label to 'Perimeter' with font size 14
plt.xlabel('Perimeter', fontsize=14);
# Set the y-axis label to 'Area' with font size 14
plt.ylabel('Area', fontsize=14);
```

```python
# Scale the data
# Initialize a StandardScaler object to standardize features by removing mean and scaling to unit variance
scaler = StandardScaler()
# Transform the rock data using standardization and create a new DataFrame
# fit_transform() calculates mean and standard deviation, then applies the transformation
# This ensures all features have mean=0 and standard deviation=1 for PCA analysis
rock = pd.DataFrame(
    scaler.fit_transform(rock), columns=['Area', 'Perimeter', 'Shape', 'Permeability']
)
```

```python
# Initialize and fit a PCA model on the rock data
# Create a PCA object that will extract 4 principal components from the standardized rock data
# n_components=4 specifies that we want to reduce the 4-dimensional data to 4 principal components
pcaModel = PCA(n_components=4);
# Fit the PCA model to the standardized rock data to learn the principal component structure
# This calculates the eigenvectors and eigenvalues from the covariance matrix
pcaModel.fit(rock);
```

```python
# Display the components
# Show the principal component loadings matrix (eigenvectors)
# Each row represents a principal component, and each column represents a feature
# These values indicate how much each original feature contributes to each principal component
pcaModel.components_
```

```python
# Display the explained variance (eigenvalues)
# Show the eigenvalues which represent the amount of variance explained by each principal component
# Larger eigenvalues indicate that the component captures more variance in the data
# This helps determine which components are most important for explaining data variability
pcaModel.explained_variance_
```

```python
# Show the factor loadings
# Calculate the factor loadings by multiplying the component loadings by the square root of eigenvalues
# Transpose the components matrix to get features as rows and components as columns
# Multiply by sqrt(explained_variance_) to scale the loadings appropriately
# Factor loadings show the correlation between original variables and principal components
pcaModel.components_.T * np.sqrt(pcaModel.explained_variance_)
```

```python
# Create a scree plot
# Define the x-axis tick positions for the scree plot (0 through 4)
xint = range(0, 5)
# Set the x-axis ticks to display at the specified integer positions
plt.xticks(xint)
# Plot the explained variance (eigenvalues) for each of the 4 principal components
# [1, 2, 3, 4] represents the component numbers, 'b*-' creates a blue line with star markers
plt.plot([1, 2, 3, 4], pcaModel.explained_variance_, 'b*-')
# Set the x-axis label to 'Factors' with font size 14 to indicate principal components
plt.xlabel('Factors', fontsize='14');
# Set the y-axis label to 'Eigenvalues' with font size 14 to show variance explained
plt.ylabel('Eigenvalues', fontsize='14');
```
